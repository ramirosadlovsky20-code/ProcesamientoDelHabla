### **Punto 2**

*Chatbot utilizando TF-IDF + la similitud del cosceno*

```python
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import spacy

# extraigo solo las preguntas de mi dataset
preg = [p['pregunta'] for p in dataset_bancario]

# cargo modelo para obtener los stopwords y poder filtrarlos
nlp = spacy.load('es_core_news_md')
stopwords_espanol = list(nlp.Defaults.stop_words)

vector_tfidf = TfidfVectorizer(stop_words=stopwords_espanol)
matriz_tfidf = vector_tfidf.fit_transform(preg)

def chatbot_tfidf(consulta, umbral = 0.3):
  # transformo la consulta del usuario
  consulta_vec = vector_tfidf.transform([consulta])

  # calculo la similitud del coseno, entre la consulta y las preguntas del detaset
  similitud = cosine_similarity(consulta_vec, matriz_tfidf)[0]

  # obtengo el indice de mayor similitud
  max_similitud_tfidf= np.argmax(similitud)
  max_score_tfidf= similitud[max_similitud_tfidf]

  if max_score_tfidf >= umbral:
    return dataset_bancario[max_similitud_tfidf]['respuesta'], max_score_tfidf, preg[max_similitud_tfidf]
  else:
    print('Disculpa, no entendi tu consulta, la podrias escribir nuevamente?', max_score_tfidf, None)


```

*Chatbot utilizando Embeddings con spaCy*

```python
import spacy

nlp = spacy.load('es_core_news_md')

# Funcion para quitar los stopwords y signos de puntuacion
def doc_limpio(texto):
  doc = nlp(texto)
  # me quedo solo con la palbra raiz
  palabras = [t.lemma_.lower() for t in doc if not t.is_stop and not t.is_punct and t.text.strip()]

  return nlp(" ".join(palabras))

# pre-proceso la preguntas del dataset con spacy
docs_preg = [doc_limpio(p['pregunta']) for p in dataset_bancario]

def chatbot_embeddings(consulta, umbral= 0.2):
  doc_consulta = doc_limpio(consulta)

  # con doc.similarity() calculo la similitud del coseno entre los vectores
  similitudes = [doc_consulta.similarity(doc) for doc in docs_preg]

  max_similitud = np.argmax(similitudes)
  max_score = similitudes[max_similitud]

  if max_score >= umbral:
    return dataset_bancario[max_similitud]['respuesta'], max_score, preg[max_similitud]
  else:
    print('Disculpa, no entendi tu consulta, la podrias escribir nuevamente?', max_score, None)
```

```python
preguntas_prueba = [
    "¿Cómo puedo ver la plata que tengo en mi caja de ahorro?",
    "Me robaron el plástico del débito, ayuda",
    "¿Cuál es el horario de atención?",
    "¿Cómo hago para hacer una torta de chocolate?",
    "Quiero saber la clave PIL del cajero",
    "¿Cómo puedo abrir una cuenta de banco?",
    "¿Cómo blanqueo mi clave?",
    "Quiero trabajar como atencion al cliente?"
]

print("=== PRUEBAS DE FUNCIONAMIENTO ===\n")
for q in preguntas_prueba:

  try:
    resp_tf, score_tf, match_tf = chatbot_tfidf(q)
    print(f"Pregunta Usuario: '{q}'")
    print(f"  [TF-IDF] Macheo con: '{match_tf}' (Score: {score_tf:.2f})")
    print(f"  [TF-IDF] Respuesta: {resp_tf[:70]}...")

  except Exception as e:
    print(f" La pregunta del Usuario: '{q}'")
    #print(f'error {e}')
    print("-" * 60)
  try:
    resp_emb, score_emb, match_emb = chatbot_embeddings(q)
    print(f"Pregunta Usuario: '{q}'")
    print(f"  [Embeddings] Macheo con: '{match_emb}' (Score: {score_emb:.2f})")
    print(f"  [Embeddings] Respuesta: {resp_emb[:70]}...")
    print("-" * 60)

  except Exception as e:
    print(f" La pregunta del Usuario: '{q}'")
    print(f'error {e}')
    print("-" * 60)
```
